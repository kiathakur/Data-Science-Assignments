```python
import pandas as pd
```

```python
# 1.Loading datasets
patient_df=pd.read_csv("Patient_Data.csv")
billing_df=pd.read_csv("Billing_Data.csv")
```

```python
# show summary with info()
patient_df.info()
```

```python
billing_df.info()
```

```python
# 2.Select only the columns relevant for billing: ['PatientID', 'Department', 'Doctor', 'BillAmount'].
patient_df[['PatientID', 'Department', 'Doctor', 'BillAmount']]
```

```python
# 3.	Drop administrative columns like ['ReceptionistID', 'CheckInTime'].
patient_df.drop(columns=['ReceptionistID', 'CheckInTime'],inplace =True)
patient_df
```

```python
#4.	Use groupby to find total bill amount per department
patient_df.groupby('Department')['BillAmount'].sum()
```

```python
#5.	Remove duplicate patient records based on PatientID.
```

```python
# Checking for duplicates
patient_df.duplicated().sum()
```

```python
# looking for duplicated value
patient_df[patient_df.duplicated()]
```

```python
# removing duplicated value
patient_df.drop_duplicates(subset='PatientID',inplace=True)
```

```python
# checking for ,duplicates value removed or not
patient_df.duplicated().sum()
```

```python
# 6.	Fill missing BillAmount values with the mean bill amount.
mean_bill=patient_df['BillAmount'].mean()
patient_df.fillna({'BillAmount': mean_bill})
patient_df
```

```python
#7.	Merge the billing dataset with patient dataset on PatientID.
merge_df=pd.merge(patient_df,billing_df,on='PatientID',how='left') # using leftmerge
merge_df
```

```python
#8.	Concatenate an additional DataFrame that contains new patients for the current week (row-wise).
                                                                
```

```python
# Creation of an additional DataFrame 
new_patient={'PatientID':[106,107],'Name':['Raj','Sam'],'Department':['Cardiology','Orthopedics'],'Doctor':['Dr. Smith','Dr.Lee'],
             'BillAmount':[5000,7000]}
new_patient_df=pd.DataFrame(new_patient)
new_patient_df
```

```python
# Concatenate (row-wise,axis=0).
concate_df=pd.concat([patient_df,new_patient_df],ignore_index=True,axis=0)
concate_df
```

```python
#9.	Concatenate new billing category columns like ['InsuranceCovered', 'FinalAmount'] (column-wise).
new_billing={'InsuranceCovered':[2000,1000,1500,3000,4000,2500,3500],'FinalAmount':[3000,5233.33,6000,4700,2233.33,2500,3500]}
new_billing_df=pd.DataFrame(new_billing)
Final_bill_df=pd.concat([concate_df,new_billing_df],axis=1)
Final_bill_df
```

```python
# Doctor wise Final Amount
Final_bill_df.groupby('Doctor')['FinalAmount'].sum()
```
