Descriptive Analytics for Numerical Columns

```python
import pandas as pd #importing pandas library
```

```python
df=pd.read_csv("sales_data_with_discounts.csv") # reading the dataset
```

```python
# 	Identify numerical columns in the dataset.
# numercial colums are the ones with int(Volume, Avg Price ,Total Sales Value ) and float data types (Discount Rate, Discount Amount and Net Sales Value)
```

```python
df.info()
```

```python
num_cols=df.select_dtypes(include=['int64','float64'])## identify numerical columns
num_cols.head()
```

```python
# Calculating mean for the numerical columns
# Mean(average)= Sum of values/number of values
mean_values=num_cols.mean() 
print(mean_values)
```

```python
# Calculating median(Middle value)
median_values=num_cols.median()
print(median_values)
```

```python
# Calculating mode-Most frequent value
mode_values=num_cols.mode().iloc[0]
print(mode_values)
```

```python
# Calculating standard deviation- measures how spread out values in a dataset are from mean
std_values=num_cols.std()
print(std_values)
```

Data Visualization

Objective: To visualize the distribution and relationship of numerical and categorical variables in the dataset.

```python
# HISTOGRAMS for each numerical columns
# Analyze the distribution (e.g., skewness, presence of outliers) and provide inferences.
```

```python
import matplotlib.pyplot as plt 
import seaborn as sns 
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
import matplotlib.colors as cs
```

```python
# Histogram of volume
sns.histplot(data=df,x='Volume',bins=15)
plt.xlabel('Volume',color='green',size=12)
plt.ylabel('Frequency',color='green',size=12)
plt.title('Histogram of Volume',color='red',size=12)
```

Histogram of Volume

Shape: Right-skewed (positively skewed)

Observation: Most transactions involve low volumes (1–6 units)

Outliers: A few orders have very high volumes (20+ units)

Inference: Sales are dominated by small-quantity purchases, but bulk orders occasionally occur and significantly increase total sales.

```python
# Histogram of Avg Price
sns.histplot(data=df,x='Avg Price',bins=15)
plt.xlabel('Avg Price',color='green',size=12)
plt.ylabel('Frequency',color='green',size=12)
plt.title('Histogram of Avg Price',color='red',size=12)
```

Avg Price

Shape: Highly right-skewed

Observation: Majority of products are low-priced

Outliers: Few very high-priced premium items

Inference: The product mix ranges from low-value to premium offerings. 

```python
# Histogram of Total Sales Value
sns.histplot(data=df,x='Total Sales Value',bins=15)
plt.xlabel('Total Sales Value',color='green',size=12)
plt.ylabel('Frequency',color='green',size=12)
plt.title('Histogram of Total Sales Value',color='red',size=12)
```

Total Sales Value

Shape: Strong right skew

Observation: Most orders generate modest revenue

Outliers: A small number of very large sales

Inference: Revenue is not evenly distributed — large-value customers play a critical role.

```python
# Histogram of Discount Rate (%)
sns.histplot(data=df,x='Discount Rate (%)',bins=15)
plt.xlabel('Discount Rate (%)',color='green',size=12)
plt.ylabel('Frequency',color='green',size=12)
plt.title('Histogram of Discount Rate (%)',color='red',size=12)
```

Discount Rate (%)

Observation: Most discounts fall between 15–20%

Inference: Discount policy is consistent.

```python
# Histogram of Discount Amount
sns.histplot(data=df,x='Discount Amount',bins=15)
plt.xlabel('Discount Amount',color='green',size=12)
plt.ylabel('Frequency',color='green',size=12)
plt.title('Histogram of Discount Amount',color='red',size=12)
```

Bar Chart Analysis for Categorical Column

```python
# 	Identify categorical columns in the dataset.
```

```python
cat_cols=df.select_dtypes(include=['object'])## identify categorical columns
cat_cols.head()
```

```python
# 	Create bar charts to visualize the frequency or count of each category.
```

```python
# Bar Chart of Date
date_counts = df["Date"].value_counts() # Counts how many times each date appears in the column and stores it as a Series.
plt.bar(date_counts.index.astype(str), date_counts.values) # X-axis : unique dates, Y-axis : count of records per date
plt.xlabel('Date',color='green')
plt.ylabel('Count',color='green')
plt.title('Bar Chart of Date',color='yellowgreen',size=15)
plt.xticks(rotation=45)
plt.show()
```

Date :Each date appears the same number of times.Hence data is evenly distributed across days.

```python
# Bar Chart of Day
day_counts = df["Day"].value_counts()
plt.bar(day_counts.index.astype(str), day_counts.values)
plt.xlabel('Day',color='green')
plt.ylabel('Count',color='green')
plt.title('Bar Chart of Day',color='yellowgreen',size=15)
plt.xticks(rotation=45)
plt.show()
```

Day :Thursday has the highest count.
Other weekdays are evenly represented.

```python
# Bar Chart of City
city_counts = df["City"].value_counts()
plt.bar(city_counts.index.astype(str), city_counts.values)
plt.xlabel('City',color='green')
plt.ylabel('Count',color='green')
plt.title('Bar Chart of City',color='yellowgreen',size=15)
plt.xticks(rotation=45)
plt.show()
```

City
Only one city appears , this dataset is city-specific.

```python
# Bar Chart of BU
BU_counts = df["BU"].value_counts()
plt.bar(BU_counts.index.astype(str), BU_counts.values)
plt.xlabel('BU',color='green')
plt.ylabel('Count',color='green')
plt.title('Bar Chart of BU',color='yellowgreen',size=15)
plt.xticks(rotation=45)
plt.show()
```

All are equally represented.

```python
# Bar Chart of Brand
Brand_counts = df["Brand"].value_counts()
plt.bar(Brand_counts.index.astype(str), Brand_counts.values)
plt.xlabel('Brand',color='green')
plt.ylabel('Count',color='green')
plt.title('Bar Chart of Brand',color='yellowgreen',size=15)
plt.xticks(rotation=45)
plt.show()
```

Conclusion 
# Summarize the key findings from the descriptive analytics and data visualizations.

Numerical variables such as Volume, Sales Value, and Discount Amount show positive skewness, meaning most transactions are small, with a few large orders contributing heavily to total sales.

Average Price indicates the presence of both low-priced and premium products, with premium items forming the higher end of the distribution.

Discount Rate remains fairly consistent across transactions, suggesting a structured and controlled discount policy.

Boxplots confirmed meaningful outliers, representing genuine high-value sales rather than data errors.

Categorical analysis shows transactions are evenly distributed across dates and business units, indicating steady operational activity.

Some brands and product categories occur more frequently, reflecting higher customer preference and demand.

Overall, business revenue is driven by many low-value sales supported by fewer high-value transactions, highlighting the importance of key customers and bulk orders.

These insights can help in pricing strategy, customer targeting, and sales forecasting for better decision-making.
