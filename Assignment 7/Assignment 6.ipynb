HYPOTHESIS TESTING

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import warnings
warnings.filterwarnings('ignore')
```

```python
# H0(Null Hypothesis): There is no increase in weekly operating cost 
# Ha(Alternate Hypothesis): There is an increase in weekly operating cost
```

```python
# Given Data
x_bar=3050 # sample mean
X=600 # number of units
n=25 # sample size
mu=4000 #= theoretical mean weekly cost according to the cost model (W = $1,000 + $5X for X = 600 units)
sigma= 125 # 5 * 25 ,theoretical standard deviation
```

```python
z_statistic=(x_bar-mu)/(sigma/np.sqrt(n))
z_statistic
```

```python
# Calculations
alpha=0.05
CI=0.95
area= alpha/2 +CI
percent_area =area*100
percent_area
```

```python
#Calculating p_value
p_value=1-stats.norm.cdf(97.5,loc=4000,scale=125) # (percent_area,loc=mu,scale=sigma
p_value
```

```python
# since p_value(1)>alpha(0.05) we accept null hypothesis -H0
```

```python
# Calculating z critical value
z_critical=stats.norm.ppf(0.975)
z_critical
```

```python
#  Since z_critical(1.95) > z_statistic(-38) ,we fail to reject the null hypothesis.
# There is no increase in weekly operating cost.
```

```python
# a) Build 99% Confidence Interval Using Sample Standard Deviation
# Assuming the sample is representative of the population, construct a 99% confidence interval for the mean number of characters printed before the print-head fails using the sample standard deviation. Explain the steps you take and the rationale behind using the t-distribution for this task.
```

```python
# Data in million of charcaters
data =np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
```

```python
# sample size
n=len(data)
n
```

```python
# Sample mean
x_bar=data.mean()
x_bar
```

```python
# Sample Standard Deviation
s=np.std(data)
s
```

```python
# Calculations
alpha=0.01 
CI=0.99
area =CI+alpha/2
df= n-1
print(area,df)
```

```python
t_critical=stats.t.ppf(0.995,14)
t_critical
```

```python
# Margin_of_error
Margin_of_error= t_critical*(s/np.sqrt(n))
Margin_of_error
```

```python
sample_mean_lower= x_bar-Margin_of_error
sample_mean_upper= x_bar+Margin_of_error
print(sample_mean_lower,sample_mean_upper)
```
