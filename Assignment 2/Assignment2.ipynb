```python
# 1.Create a list of 5 integers. 
#Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.
```

```python
List1=[10,20,15,25,55,60,80] ### creation of a list
print(List1)
```

```python
List1.append(100) ### adds element at the end of the list
print("After append :",List1)
```

```python
List1.extend([70,80,90])### adds multiple elements to the end of an existing list
print("After extend:",List1)
```

```python
List1.insert(4,30)### inserts element at a given index position 
print("After insert(30 at index 4):",List1)
```

```python
List1.remove(100)### delets the given element from the list
print("After remove(100):",List1)
```

```python
List1.pop()### delets the last element of the given list
print("After pop:",List1)
```

```python
index_of_55=List1.index(55)## gives index position of given element
print("index of 55:",index_of_55)
```

```python
Count_80=List1.count(80)### gives the number of times given element is present in the list
print("Count of 80:",Count_80)
```

```python
List1.sort()### sorts list in ascending order
print("After sort:",List1)
```

```python
List1.reverse()### reverses the order 
print("After reverse:",List1)
```

```python
List1.clear()## deletes every item in the list
print("After clear:",List1)
```

```python
#11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
# Use .info(), .describe()
# Select 'Name' and 'City' columns
# Drop 'City' column
# Fill any missing values in 'Salary' column with the mean
# Remove any duplicate rows
```

```python
# Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'
import pandas as pd
import numpy as np
data={"Name":["Kiraan","Nishad","Prachi","Ruhi","Satej","Tejit","Bhaumi","Purvi","Kiaan","Ninad"],"Age":[32,35,42,37,38,33,28,45,41,29],
      "City":["Mumbai","Mumbai","Mumbai","Pune","Thane","Panvel","Pune","Palghar","Chennai","Bangalore"],
      "Salary":[50000,62000,55000,45000,46000,42000,43000,40000,35000,38000]}
df1=pd.DataFrame(data)
print(df1)
```

```python
df1.info() # gives information such as column name,number of enteries,null/non-null count and datatype
```

```python
df1.describe() ## generates summary statistics for the numeric columns in dataframe
```

```python
df1[["Name","City"]]# Select 'Name' and 'City' columns
```

```python
# Drop 'City' column
df1_dropped=df1.drop(columns=["City"])
print(df1_dropped)
```

```python
# Fill any missing values in 'Salary' column with the mean
data1={"Name":["Kiraan","Nishad","Prachi","Ruhi","Satej","Tejit","Bhaumi","Purvi","Kiaan","Ninad"],"Age":[32,35,42,37,38,33,28,45,41,29],
      "City":["Mumbai","Mumbai","Mumbai","Pune","Thane","Panvel","Pune","Palghar","Chennai","Bangalore"],
      "Salary":[50000,62000,55000,45000,46000,42000,43000,np.nan,40000,38000]}
df2=pd.DataFrame(data1)
df2
```

```python
mean_Salary=df2['Salary'].mean()
df2.fillna({'Salary': mean_Salary}, inplace=True) ### replaces missing value with mean
(df2)
```

```python
## Creating dataframe with duplicate row
data2={"Name":["Kiraan","Nishad","Prachi","Ruhi","Satej","Tejit","Bhaumi","Purvi","Purvi","Ninad"],"Age":[32,35,42,37,38,33,28,45,45,29],
      "City":["Mumbai","Mumbai","Mumbai","Pune","Thane","Panvel","Pune","Palghar","Palghar","Bangalore"],
      "Salary":[50000,62000,55000,45000,46000,42000,43000,40000,40000,38000]}
df3=pd.DataFrame(data2)
df3.duplicated().sum() ## checking duplicate values
```
