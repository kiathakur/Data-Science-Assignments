```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
from sklearn.metrics.pairwise import cosine_similarity
```

```python
#Load Dataset
df = pd.read_csv("anime.csv")
df.head()
```

```python
df.tail()
```

```python
# converting any HTML character entities within each string back into their original, readable characters. 
import html
df['name']=df['name'].apply(html.unescape)
```

```python
df.shape
```

```python
# remove leading and trailing whitespace, and double quotes from all strings in that column. 
df['name']=df['name'].str.strip().str.strip('"')
```

```python
# Creatng pivot table
df1= df.pivot_table(index= 'name',columns='members',values='rating')
df1.fillna(0,inplace=True)
df1.head()
```

```python
# Calculating cosine similarity score
cosine=cosine_similarity(df1)
```

```python
df1.index
```

```python
def recommend_anime(anime_name):
    if anime_name in df1.index:
        index = df1.index.get_loc(anime_name)

        similar = sorted(
            list(enumerate(cosine[index])),
            key=lambda x: x[1],
            reverse=True
        )[1:6]

        print(f"Recommended anime for {anime_name} are:")
        print("*" * 40)

        for anime in similar:
            print(df1.index[anime[0]])
    else:
        print(f"{anime_name} anime is not in the list")
```

```python
recommend_anime(df1.index[2])
```

```python
recommend_anime('Üks Uks')
```

```python
recommend_anime('Eiji')
```

```python
def recommend_anime(anime_name, top_n=5, threshold=0.3):
    if anime_name not in df1.index:
        print(f"{anime_name} anime is not in the list")
        return

    index = df1.index.get_loc(anime_name)

    # Get similarity scores
    sim_scores = list(enumerate(cosine[index]))

    # Sort by similarity score
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # Remove itself and apply threshold
    sim_scores = [
        anime for anime in sim_scores[1:]
        if anime[1] >= threshold
    ]

    print(f"Recommended anime for {anime_name} (threshold={threshold}):")
    print("*" * 50)

    for anime in sim_scores[:top_n]:
        print(f"{df1.index[anime[0]]}")
```

```python
recommend_anime('vivi', top_n=5, threshold=0.2)
```

```python
recommend_anime('Eiji', top_n=5, threshold=0.2)
```

```python
recommend_anime('Eiji', top_n=5, threshold=0)
```

```python
Interview Questions:
1. Can you explain the difference between user-based and item-based collaborative filtering?
2. What is collaborative filtering, and how does it work?

1. Difference Between User-Based and Item-Based Collaborative Filtering
User-Based Collaborative Filtering:

Finds users with similar tastes
Recommends items liked by similar users
Example
If User A and User B both liked Naruto and Bleach,
and User B liked One Piece,
→ recommend One Piece to User A.

Pros
Personalized recommendations

Cons
Poor scalability with many users
User preferences change frequently

Item-Based Collaborative Filtering:

Finds items that are similar
Recommends items similar to what the user already liked

Example
If Naruto and Bleach are often liked by the same users,
→ recommend Bleach to someone who liked Naruto.

Pros
More stable
Better scalability
Used by Netflix, Amazon

Cons
Less personalized than user-based

2. What is Collaborative Filtering and How Does It Work?
Definition

Collaborative filtering is a recommendation technique that uses user behavior and interactions and recommends items based on collective preferences.
It does not require item metadata (genre, type, etc.).

How It Works
1.Collect user–item interaction data (ratings, likes, views)
2.Create a user–item matrix
3.Measure similarity (cosine, Pearson, etc.)
4.Identify similar users or items
5.Recommend items based on similarity patterns
```
