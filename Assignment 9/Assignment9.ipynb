```python
# Check missing values
df.isnull().sum()
```

```python
# identifying categorical columns
categorical_cols=df.select_dtypes(include=['object']).columns
categorical_cols
```

```python
# Checking for missing values '?'in categorical columns
(df[categorical_cols] == ' ?').sum()
```

```python
# percentage of missing values
missing_summary = pd.DataFrame({
    'Missing_Count': (df[categorical_cols] == ' ?').sum(),
    'Missing_Percentage': (df[categorical_cols] == ' ?').mean() * 100
})
missing_summary
```

```python
# Replace '?' with NaN
df.replace(" ?", np.nan, inplace=True)
```

```python
#Impute Categorical Columns with Mode
for col in df.select_dtypes(include="object").columns:
    df[col].fillna(df[col].mode()[0], inplace=True) # [0] selects the first most frequent value
```

```python
# Checking for missing values replaced
(df[categorical_cols] == ' ?').sum()
```

```python
# selecting numerical columns
num_features = df.select_dtypes(include=["int64", "float64"]).columns
```

```python
# a. Standard Scaling
from sklearn.preprocessing import StandardScaler
```

```python
std_scaler = StandardScaler()
df_std_scaled = df.copy()
df_std_scaled[num_features] = std_scaler.fit_transform(df[num_features])
```

```python
df_std_scaled[num_features]
```

```python
# b. Min-Max Scaling
from sklearn.preprocessing import MinMaxScaler
```

```python
mm_scaler = MinMaxScaler()
df_mm_scaled = df.copy()
df_mm_scaled[num_features] = mm_scaler.fit_transform(df[num_features])
```

```python
# applying onehotencoder and converting to dataframe
df1=pd.DataFrame(one_hot.fit_transform(df[['sex']]).toarray(),columns=['Female','Male'])
df2=pd.DataFrame(one_hot.fit_transform(df[['race']]).toarray(),columns=['White','Black','Asian-Pac-Islander', 'Amer-Indian-Eskimo', 'Other'])
df3=pd.DataFrame(one_hot.fit_transform(df[['income']]).toarray(),columns=['<=50K','>50K'])
```
